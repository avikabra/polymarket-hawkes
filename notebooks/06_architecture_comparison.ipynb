# Notebook 06: Architecture Comparison — 4×3 R² Matrix

**Central thesis result.** Loads `results/metrics_all.parquet` and plots the architecture × category R²_OOS matrix with bootstrap CI error bars.

```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
# Load and pivot to 4×3 matrix (architecture × category)
results_path = Path("results/metrics_all.parquet")
if results_path.exists():
    df = pd.read_parquet(results_path)
    # Use shock embedding, drop 'all' category for 3-category axis
    shock = df[(df['embedding_type'] == 'shock') & (df['category'] != 'all')]
    matrix = shock.pivot(index='architecture', columns='category', values='r2_oos')
    # Enforce row order: linear, lstm, tcn, transformer
    arch_order = ['linear', 'lstm', 'tcn', 'transformer']
    matrix = matrix.reindex([a for a in arch_order if a in matrix.index])
    print(matrix.to_string())
else:
    print("results/metrics_all.parquet not found — run script 20 first")
    matrix = None
```
